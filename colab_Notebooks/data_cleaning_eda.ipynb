##SECTION 1 - Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("cardekho_dataset.csv.zip")
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
print("null values:", df.isnull().sum())
print("duplicate values:",df.duplicated().sum())
```

```python
df.head()
```

```python
df["seats"].value_counts().sort_index()
```

```python
df[df["seats"]==0]
df=df[df["seats"]!=0]
df["seats"].value_counts().sort_index()
```

```python
df.dtypes
```

```python
df["selling_price"].describe()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=df,x="selling_price",bins=50,kde=True)
plt.title("Distribution of selling price")
plt.xlabel("Selling price")
plt.ylabel("Number of cars")
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.boxplot(x=df["selling_price"])
plt.title("selling price - Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,x="fuel_type")
plt.title("number of cars by fuel")
plt.xlabel("fuel_type")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df,x="transmission_type")
plt.title("Cars by transmissionn")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="seller_type")
plt.title("cars by seller type")
plt.show()
```

```python
brand_counts=df["brand"].value_counts().head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=brand_counts.values,y=brand_counts.index)
plt.title("Top 10 cars brands by number of listing")
plt.xlabel("number  of cars")
plt.ylabel("brand")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,x="vehicle_age",y="selling_price")
plt.title("selling price vs vechicle age")
plt.xlabel("Vehicle age")
plt.ylabel("selling price")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,x="km_driven",y="selling_price")
plt.title("selling price vs km driven")
plt.xlabel("km driven")
plt.ylabel("selling price")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,x="mileage",y="selling_price")
plt.title("selling price vs mileage")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,x="engine",y="selling_price")
plt.title("Engine size vs selling price")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,x="max_power",y="selling_price")
plt.title("horespower vs selling price")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.boxplot(data=df,x="fuel_type",y="selling_price")
plt.title("selling price by fuel type")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="transmission_type",y="selling_price")
plt.title("selling price by transmission type")
plt.show()
```

```python
brand_price=(df.groupby("brand")["selling_price"].mean().sort_values(ascending=False))
plt.figure(figsize=(10,5))
sns.barplot(x=brand_price.values,y=brand_price.index)
plt.title(" top 10 brand by Average selling ")
plt.xlabel("Average selling price")
plt.ylabel("brand")
plt.show()

```

```python
numerical_column=["selling_price","km_driven","mileage","engine","max_power","seats",]
corr=df[numerical_column].corr()
```

```python
print(corr)
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("correlation matrix")
plt.show()
```
